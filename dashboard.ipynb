```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Load data
df = pd.read_csv("sales_data.csv")
df.head()
```

```python
df.isnull().sum()

df['Date'] = pd.to_datetime(df['Date'])
```

```python
sales_over_time = df.groupby('Date')['Total_Sales'].sum()

plt.figure(figsize=(8,4))
sales_over_time.plot()
plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("visualizations/sales_trend.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='Product', y='Total_Sales', data=df)
plt.title("Sales by Product")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("visualizations/Product_sales.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Product', y='Price', data=df)
plt.title("Price Distribution by Product")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("visualizations/price_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['Price','Quantity','Total_Sales']].corr(),
            annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("visualizations/heatmap.png")
plt.show()
```
